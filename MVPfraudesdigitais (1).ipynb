```python
# Célula 2: Treinamento do Modelo de Detecção de Fraude (Simulação Prática)
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline

# Dataset simulado de mensagens de fraudes digitais / golpes comuns
dados_treino = {
    'texto': [
        "Seu token de acesso expirou, clique aqui para atualizar sua conta imediatamente.",
        "Parabéns! Você ganhou um sorteio de 5000 reais, clique no link para resgatar.",
        "Olá pai, troquei de número, me manda um PIX urgente para pagar uma conta.",
        "Reunião de alinhamento de projeto marcada para amanhã às 14h.",
        "Segue em anexo o relatório mensal de atividades da diretoria.",
        "Lembrete: sua fatura vencimento dia 10 está disponível no aplicativo do banco.",
        "URGENTE: Bloqueio preventivo no seu CPF. Regularize agora via link.",
        "Favor confirmar o recebimento do e-mail com a ata da última reunião."
    ],
    'rotulo': [
        'Fraude / Golpe', 'Fraude / Golpe', 'Fraude / Golpe',
        'Legítimo', 'Legítimo', 'Legítimo',
        'Fraude / Golpe', 'Legítimo'
    ]
}

df = pd.DataFrame(dados_treino)

# Criando o pipeline de Machine Learning (Vetorização + Classificador)
# Nota: Para rigor de mestrado, isso valida o conceito de aprendizado de máquina aplicado a fraudes
modelo = make_pipeline(TfidfVectorizer(), LogisticRegression())
modelo.fit(df['texto'], df['rotulo'])

print("Modelo treinado com sucesso para o MVP!")
```

```python
# Célula Única: Treinamento + Apresentação + Interface Corrigida
import pandas as pd
import gradio as gr
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline

# 1. Dataset Simulado (Base do MVP)
# Certifique-se de ter mensagens suficientes e variadas para o treinamento
dados_treino = {
    'texto': [
        "Seu token de acesso expirou, clique aqui para atualizar.",
        "Parabéns! Você ganhou um sorteio de 5000 reais.",
        "Olá pai, troquei de número, me manda um PIX urgente.",
        "Reunião de alinhamento amanhã às 14h.",
        "Segue em anexo o relatório mensal de atividades.",
        "Lembrete: sua fatura vence dia 10.",
        "URGENTE: Bloqueio preventivo no seu CPF.",
        "Favor confirmar o recebimento do e-mail com a ata.",
        "Oportunidade única de investimento com retorno garantido.",
        "Confirmação de agendamento de consulta médica."
    ],
    'rotulo': [
        'Fraude / Golpe', 'Fraude / Golpe', 'Fraude / Golpe',
        'Legítimo', 'Legítimo', 'Legítimo',
        'Fraude / Golpe', 'Legítimo', 'Fraude / Golpe', 'Legítimo'
    ]
}
df = pd.DataFrame(dados_treino)

# 2. Treinamento do Modelo (Recriando o pipeline)
modelo = make_pipeline(TfidfVectorizer(), LogisticRegression())
modelo.fit(df['texto'], df['rotulo'])
print("Modelo treinado com sucesso!")

# 3. Lógica de Predição
def prever_fraude(mensagem):
    if not mensagem.strip():
        return "Por favor, digite uma mensagem para análise."
    predicao = modelo.predict([mensagem])[0]
    probabilidades = modelo.predict_proba([mensagem])[0]
    confianca = max(probabilidades) * 100
    return f"Classificação: {predicao}\nConfiança do Modelo: {confianca:.2f}%"

# 4. Textos de Apresentação Acadêmica
texto_intro = """
# MVP - Detector de Fraudes Digitais
* **Disciplina:** Tópicos em Segurança Cibernética 4 (Redes Neurais Aplicadas à Segurança)
* **Objetivo:** Protótipo funcional em ambiente de nuvem (Google Colab) para classificação automática de mensagens maliciosas e engenharia social utilizando aprendizado de máquina.
"""

# 5. Interface Web com Gradio
demo = gr.Interface(
    fn=prever_fraude,
    inputs=gr.Textbox(lines=3, label="Cole a mensagem para análise:"),
    outputs=gr.Textbox(label="Resultado da Análise de Fraude"),
    title="MVP: Detector de Fraudes Digitais",
    description=texto_intro,
    submit_btn=gr.Button("Enviar", variant="primary"),
    clear_btn="Limpar"
)

# Rodar a aplicação
demo.launch(share=True)
```
